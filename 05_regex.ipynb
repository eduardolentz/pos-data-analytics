```python
import re
```

```python
match = re.match('Antonio', 'Rua Antonio Basilio')
print(match)
```

```python
search = re.search('Antonio', 'Rua Antonio Basilio')
print(search)
```

```python
findall = re.findall('div','<div class=conatiner> <div class=barra >')
print(findall)
```

```python
split = re.split('c','Descomplica')
print(split)
```

```python
sub = re.sub('uma plataforma', 'a melhor plataforma', 'A Descomplica é uma plataforma de ensino online')
print(sub)
```

```python
p = re.compile(r'[0-9]+')
print(p.findall('12 campos de futebol, 10 bolas, 25 chuteiras'))
```
