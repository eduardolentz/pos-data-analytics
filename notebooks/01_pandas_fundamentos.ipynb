```python
vendas = {
    "chocolate":[5,3,1,2,2],
    "marshmallow":[2,1,0,8,3],
    "bala":[1,2,4,6,4]
}
print(vendas)
```

```python
import pandas as pd

compras = pd.DataFrame(vendas)

compras = pd.DataFrame(vendas, index=['Matheus','Larissa','David','Cris','Sabrina'])
#compras.loc['Larissa']

# Adicionando nova coluna
compras['total'] = compras['chocolate'] + compras['marshmallow'] + compras['bala']
compras

# Renomeando coluna
compras.rename(columns={'chocolate':'col1','marshmallow':'col2','bala':'col3','total':'col4'},inplace=True)
compras

# Excluindo coluna

compras.drop('col4',axis=1,inplace=True)
compras

# Excluindo linha

compras.drop('Larissa',axis=0,inplace=True)
compras

# Seleção de linhas com condições booleanas
compras[compras['col1']>1]
```
